```python
import anndata as ad
from pathlib import Path
import anndata as ad

root = Path("/work/magroup/skrieger/Syndri/data/datasets/agingbrain_merfish")
out = root / "interim/single-cell/rctd_donor_references"
out.mkdir(parents=True, exist_ok=True)

reference = ad.read_h5ad('/work/magroup/skrieger/Syndri/data/datasets/agingbrain_merfish/interim/single-cell/sc_integrated_agematched.h5ad')
```

```python
for donor in range(5, 9):
    donor_id = f"10X_MsBrainAgingSpatialDonor_{donor}"
    mask = reference.obs["donor_id"].eq(donor_id).to_numpy()
    if not mask.any():
        raise ValueError(f"No reference cells for {donor_id}")

    subset = reference[mask].to_memory()
    path = out / f"reference_donor_{donor}.h5ad"
    subset.write_h5ad(path)
    print(f"{path}: {subset.n_obs} cells")
```
