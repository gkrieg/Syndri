```python
import anndata as ad

sc_data = ad.read_h5ad("/work/magroup/skrieger/Syndri/data/split_sc_ref/breast_cancer_chromium.h5ad")
st_data = ad.read_h5ad("/work/magroup/skrieger/Syndri/data/split_spatial/breast_cancer_xenium.h5ad")
st_data.var['feature'] = st_data.var.index
st_data.var = st_data.var.set_index('gene_ids')
overlap = set(sc_data.var.index.tolist()).intersection(set(st_data.var.index.tolist()))
```

```python
gene_mask = sc_data.var_names.isin(overlap)
sc_data = sc_data[:,gene_mask]

gene_mask = st_data.var_names.isin(overlap)
st_data = st_data[:,gene_mask]
```

```python
counts = sc_data.obs['cell_type'].value_counts()
drop_cts = counts[counts < 25].index.tolist()

sc_data = sc_data[~sc_data.obs['cell_type'].isin(drop_cts)]
```
