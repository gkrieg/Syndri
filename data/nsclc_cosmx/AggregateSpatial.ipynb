```python
import pandas as pd
import os
spatial_dir = '/work/magroup/skrieger/Syndri/data/datasets/breast_xenium/raw/spatial/'
metadata_files = sorted([f for f in os.listdir(spatial_dir) if f.endswith('cells.parquet')])
mtx_files = sorted([f for f in os.listdir(spatial_dir) if f.endswith('feature_matrix.h5')])
```

```python
adata_fnames = ['_'.join(f.split('_')[:-1])+'.h5ad' for f in metadata_files]
adata_fnames
```

```python
import h5py 
import scanpy as sc
for i, f in enumerate(metadata_files):
    met = pd.read_parquet(os.path.join(spatial_dir,f))
    adata = sc.read_10x_h5(os.path.join(spatial_dir,mtx_files[i]))
    adata.obs = adata.obs.merge(met,left_index=True,right_on='cell_id',how='inner')

    adata.write_h5ad(os.path.join(spatial_dir,adata_fnames[i]))
    # print(X)
```

```python
import h5py 
import scanpy as sc
concat_list = []
for i, f in enumerate(metadata_files):
    met = pd.read_parquet(os.path.join(spatial_dir,f))
    adata = sc.read_10x_h5(os.path.join(spatial_dir,mtx_files[i]))
    adata.obs = adata.obs.merge(met,left_index=True,right_on='cell_id',how='inner')
    adata.obs['tissue'] = adata_fnames[i].split('.')[0]
    concat_list.append(adata.copy())
    # print(X)
```
